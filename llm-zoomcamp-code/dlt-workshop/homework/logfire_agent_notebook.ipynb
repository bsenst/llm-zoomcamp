## 4. Initialize Logfire instrumentation

Configure Logfire in the notebook and enable instrumentation for the agent workflow so spans are captured for agent runs, LLM calls, and tool calls.

```python
if os.getenv('LOGFIRE_TOKEN'):
    logfire.configure(token=os.getenv('LOGFIRE_TOKEN'), send_to_logfire=True)
elif os.getenv('LOGFIRE_READ_TOKEN'):
    logfire.configure(token=os.getenv('LOGFIRE_READ_TOKEN'), send_to_logfire=False)
else:
    logfire.configure(send_to_logfire=False)

logfire.instrument_pydantic_ai()
print('Logfire configured')
```
